```python

from pathlib import Path
import pandas as pd

# ============================================================
# 1. DEFINE ABSOLUTE FILE PATHS
# ============================================================

base_folder = Path(r"F:\AI_ATS_SCANNER")

jd_path = base_folder / "data" / "JD.txt"

input_csv = (
    base_folder
    / "data"
    / "resumes"
    / "analysis_results"
    / "resume_analysis.csv"
)

output_csv = (
    base_folder
    / "data"
    / "resumes"
    / "analysis_results"
    / "shortlisted_candidates.csv"
)


# ============================================================
# 2. READ AND DISPLAY JOB DESCRIPTION
# ============================================================

if not jd_path.exists():
    raise FileNotFoundError(f"JD not found: {jd_path}")

jd_text = jd_path.read_text(encoding="utf-8")

print("=" * 60)
print("JOB DESCRIPTION")
print("=" * 60)
print(jd_text)


# ============================================================
# 3. LOAD RESUME ANALYSIS CSV
# ============================================================

if not input_csv.exists():
    raise FileNotFoundError(
        f"CSV not found: {input_csv}\n"
        "Check whether the file was created successfully."
    )

df = pd.read_csv(input_csv, encoding="utf-8-sig")

print(f"\nTotal candidates loaded: {len(df)}")


# ============================================================
# 4. VALIDATE REQUIRED COLUMNS
# ============================================================

required_columns = [
    "file_name",
    "experience_years",
    "experience_months",
    "python",
    "sql",
    "machine_learning"
]

missing_columns = set(required_columns) - set(df.columns)

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")


# ============================================================
# 5. CONVERT REQUIRED COLUMNS TO NUMERIC
# ============================================================

numeric_columns = required_columns[1:]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove rows with missing values in required fields
df = df.dropna(subset=required_columns).copy()


# ============================================================
# 6. FILTER EXPERIENCE: 0 TO 2 YEARS
# ============================================================

df["total_experience_years"] = (
    df["experience_years"]
    + df["experience_months"] / 12
)

experience_filter = (
    (df["total_experience_years"] >= 0)
    & (df["total_experience_years"] <= 2)
)


# ============================================================
# 7. FILTER SKILLS: AT LEAST 2 OUT OF 3
# ============================================================

skill_columns = ["python", "sql", "machine_learning"]

df["skills_matched"] = df[skill_columns].eq(1).sum(axis=1)

skills_filter = df["skills_matched"] >= 2


# ============================================================
# 8. SHORTLIST CANDIDATES
# ============================================================

shortlisted_df = df[
    experience_filter & skills_filter
].copy()

shortlisted_df = shortlisted_df.sort_values(
    by=["total_experience_years", "skills_matched"],
    ascending=[True, False]
)

display(shortlisted_df)
```

```python
# ============================================================
# 9. SAVE SHORTLISTED CSV
# ============================================================

shortlisted_df.to_csv(
    output_csv,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 60)
print("SHORTLISTING SUMMARY")
print("=" * 60)

print(f"Candidates evaluated : {len(df)}")
print(f"Candidates shortlisted: {len(shortlisted_df)}")
print(f"Output saved to      : {output_csv}")


# ============================================================
# 10. DISPLAY RESULTS
# ============================================================

display(
    shortlisted_df[
        [
            "file_name",
            "experience_years",
            "experience_months",
            "total_experience_years",
            "python",
            "sql",
            "machine_learning",
            "skills_matched"
        ]
    ]
)
```
